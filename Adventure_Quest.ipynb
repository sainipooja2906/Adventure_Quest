```python
import random
import time
import os

def pause(text, delay=1):
    """Print text with a short pause for dramatic effect."""
    print(text)
    time.sleep(delay)
# Player Class
class Player:
    def __init__(self, name):
        self.name = name
        self.health = 100

    def take_damage(self, amount):
        self.health -= amount
        print(f"💥 Ouch! You lost {amount} health. Current health: {self.health}")

    def heal(self, amount):
        self.health += amount
        if self.health > 100:
            self.health = 100
        print(f"✨ You gained {amount} health. Current health: {self.health}")

# Game Class
class AdventureGame:
    def __init__(self, player):
        self.player = player

    def start(self):
        pause(f"\n🌲 Welcome, {self.player.name}, to the Adventure Quest! 🌲")
        pause("You wake up in a mysterious forest with no memory of how you got here...")
        pause("Your mission: find the hidden treasure without losing all your health!\n")
        self.intro_area()

    def intro_area(self):
        while True:
            pause("\nYou stand at the forest’s edge. Paths: NORTH (forest) or EAST (river).")
            choice = input("👉 Where do you want to go? (north/east/quit): ").lower()
            if choice == "north":
                self.forest()
            elif choice == "east":
                self.river()
            elif choice == "quit":
                print("👋 Goodbye, adventurer! Your journey ends here.")
                self.save_progress()
                break
            else:
                print("❌ Invalid choice. Try again!")

    def forest(self):
        pause("\nYou walk deeper into the forest. The air feels heavy and strange...")
        event = random.choice(["find_potion", "snake_bite", "nothing"])
        if event == "find_potion":
            pause("🍃 You find a healing herb and eat it.")
            self.player.heal(random.randint(10, 25))
        elif event == "snake_bite":
            pause("🐍 A snake bites you suddenly!")
            self.player.take_damage(random.randint(10, 25))
        else:
            pause("It’s eerily quiet... nothing happens.")

        pause("Paths ahead: NORTH (mountain) or EAST (river).")
        while True:
            choice = input("👉 Where next? (north/east/back): ").lower()
            if choice == "north":
                self.mountain()
                break
            elif choice == "east":
                self.river()
                break
            elif choice == "back":
                self.intro_area()
                break
            else:
                print("❌ Invalid choice. Try again!")

    def mountain(self):
        pause("\nYou climb up the rocky mountain path...")
        event = random.choice(["rockslide", "clear_path"])
        if event == "rockslide":
            pause("🪨 A small rockslide hits you!")
            self.player.take_damage(random.randint(10, 20))
        else:
            pause("You safely reach the top and spot a glowing cave.")

        while True:
            choice = input("🕳️ Do you want to enter the cave? (yes/no): ").lower()
            if choice == "yes":
                self.cave()
                break
            elif choice == "no":
                pause("You decide it's too risky and climb down.")
                self.forest()
                break
            else:
                print("❌ Invalid choice. Try again!")

    def river(self):
        pause("\n🌊 You reach a wide, rushing river. A weak bridge crosses it.")
        event = random.choice(["bridge_breaks", "cross_safe"])
        if event == "bridge_breaks":
            pause("😱 The bridge snaps! You fall into the river!")
            self.player.take_damage(random.randint(20, 30))
        else:
            pause("You cross safely to the other side and find a trail leading to the mountains.")
            self.mountain()

        if self.player.health <= 0:
            pause("💀 You were swept away by the current... GAME OVER 💀")
            self.play_again()

    def cave(self):
        pause("\nYou step into the glowing cave...")
        pause("A treasure chest sparkles before you.")
        event = random.choice(["treasure", "trap"])
        if event == "treasure":
            pause("✨ You open the chest and find GOLD and JEWELS! 🏆")
            pause("🎉 Congratulations, brave adventurer — YOU WIN! 🎉")
        else:
            pause("💣 A hidden trap explodes as you touch the chest!")
            self.player.take_damage(random.randint(25, 40))
            if self.player.health <= 0:
                pause("💀 You couldn’t survive the blast. Game Over!")
                self.play_again()
                return
            else:
                pause("You barely survive and crawl out of the cave...")

        self.play_again()

    def save_progress(self):
        """Save the player's progress in a file."""
        with open("savegame.txt", "w") as file:
            file.write(f"{self.player.name},{self.player.health}")
        print("💾 Game progress saved!")

    def load_progress(self):
        """Load progress from file if available."""
        if os.path.exists("savegame.txt"):
            with open("savegame.txt", "r") as file:
                name, health = file.read().split(",")
                self.player.name = name
                self.player.health = int(health)
                print(f"🔁 Loaded previous game for {self.player.name} (Health: {self.player.health})")
        else:
            print("⚠️ No previous save found. Starting a new adventure.")

    def play_again(self):
        while True:
            choice = input("\n🔁 Do you want to play again? (yes/no): ").lower()
            if choice == "yes":
                print("\nRestarting the adventure...\n")
                self.player.health = 100
                self.start()
                break
            elif choice == "no":
                print("👋 Farewell, adventurer. Until next time!")
                self.save_progress()
                break
            else:
                print("❌ Invalid choice. Try again!")

# Main Program
def main():
    print("🎮 ADVENTURE QUEST – Text-based Python Game")
    name = input("Enter your player name: ")
    player = Player(name)
    game = AdventureGame(player)

    game.load_progress()
    game.start()

if __name__ == "__main__":
    main()
```
